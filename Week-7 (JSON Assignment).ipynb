```python
import pandas as pd
import requests

api_url = "https://raw.githubusercontent.com/owid/covid-19-data/refs/heads/master/public/data/latest/owid-covid-latest.json"

raw_data = requests.get(api_url).json()

all_countries = pd.DataFrame()

for country_code in raw_data.keys():
    single_country_df = pd.json_normalize(raw_data[country_code])
    single_country_df["Country_Code"] = country_code
    all_countries = pd.concat([all_countries, single_country_df], ignore_index = True)
    
#all_countries.to_csv("all_countries.csv")

city_url = "https://raw.githubusercontent.com/dr5hn/countries-states-cities-database/refs/heads/master/json/countries%2Bcities.json"

city_data = requests.get(city_url).json()

df = pd.DataFrame(city_data)

all_countries.set_index("Country_Code", inplace= True)
df.set_index("iso3", inplace = True)

joined_df = all_countries.join(df, how = "left")

joined_df.reset_index(inplace = True)
```
